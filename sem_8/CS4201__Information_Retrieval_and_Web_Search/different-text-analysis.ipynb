# English Text Analysis

```python
from java.io import StringReader
from org.apache.lucene.analysis.tokenattributes import CharTermAttribute
from org.apache.lucene.analysis.en import EnglishAnalyzer
from org.apache.lucene.analysis.core import WhitespaceAnalyzer
from org.apache.lucene.analysis.standard import StandardAnalyzer
test = "The quick brown foxes were jumping over lazy dog's."

standardAnalyzer = EnglishAnalyzer()
'''
* This creates an instance of Lucene's EnglishAnalyzer.
* The EnglishAnalyzer tokenizes English text and applies common filters such as:
    * Lowercasing
    * Removing stop words (like "the", "and", etc.)
    * Stemming (e.g., "running" → "run")
'''

stream = standardAnalyzer.tokenStream("", StringReader(test))
stream.reset()
tokens = []
while stream.incrementToken():
    tokens.append(stream.getAttribute(CharTermAttribute.class_).toString())
print(tokens)
```

```python
from org.apache.lucene.analysis.standard import StandardAnalyzer
from org.apache.lucene.analysis.tokenattributes import CharTermAttribute
from org.apache.lucene.analysis import CharArraySet
from java.io import StringReader

# ✅ Step 1: Define custom stop words (case-insensitive)
custom_stop_words = ["the", "over", "lazy"]

# Create a Java-compatible CharArraySet from the list
stop_set = CharArraySet(len(custom_stop_words), True)  # True = ignoreCase
for word in custom_stop_words:
    stop_set.add(word)

# ✅ Step 2: Create StandardAnalyzer with the custom stop word set
analyzer = StandardAnalyzer(stop_set)

# ✅ Step 3: Provide input text
text = "The quick brown fox jumped over the lazy dog"
token_stream = analyzer.tokenStream("field", StringReader(text))

# ✅ Step 4: Extract tokens
token_stream.reset()
tokens = []

char_term_attr = token_stream.getAttribute(CharTermAttribute.class_)
while token_stream.incrementToken():
    tokens.append(char_term_attr.toString())

token_stream.end()
token_stream.close()

print(tokens)
```

# German Text Analysis

```python
import lucene

# Initialize first!
# lucene.initVM()

from org.apache.lucene.analysis.de import GermanAnalyzer
from org.apache.lucene.analysis.standard import StandardAnalyzer
from java.io import StringReader

def simple_analysis():
    german_analyzer = GermanAnalyzer()
    analyzer = EnglishAnalyzer()
    text = '''Die schönen Häuser in München gefallen mir sehr  
    Ich laufe schnell, du läufst langsam, wir sind gelaufen
    Der alte Mann geht, die alten Männer gehen
    Bücher, Büchern, Buch'''
    
    # Simple tokenization
    stream = german_analyzer.tokenStream("field", StringReader(text))
    from org.apache.lucene.analysis.tokenattributes import CharTermAttribute
    term_attr = stream.addAttribute(CharTermAttribute.class_)
    
    stream.reset()
    tokens = []
    while stream.incrementToken():
        tokens.append(term_attr.toString())
    stream.close()
    
    print("German text after GermanAnalyzer is applied:", tokens)

    stream = analyzer.tokenStream("field", StringReader(text))
    from org.apache.lucene.analysis.tokenattributes import CharTermAttribute
    term_attr = stream.addAttribute(CharTermAttribute.class_)
    
    stream.reset()
    tokens = []
    while stream.incrementToken():
        tokens.append(term_attr.toString())
    stream.close()

    print("German text after EnglishAnalyzer is applied:", tokens)
simple_analysis()
```

# Bengali Text Analysis

```python
import lucene

# Initialize PyLucene VM (uncomment when running)
# lucene.initVM()

from org.apache.lucene.analysis.bn import BengaliAnalyzer
from org.apache.lucene.analysis.standard import StandardAnalyzer
from java.io import StringReader

def compare_bengali_analyzers():
    bengali_text = '''
    বাংলা ভাষায় লেখা একটি নমুনা পাঠ্য
    আমি স্কুলে যাই এবং তুমি বাজারে যাও
    বই, বইগুলি, বইয়ের, বইগুলো
    '''
    
    analyzers = {
        "BengaliAnalyzer": BengaliAnalyzer(),
        "StandardAnalyzer": StandardAnalyzer()
    }
    
    from org.apache.lucene.analysis.tokenattributes import CharTermAttribute
    
    def analyze_text(text, analyzer_name, analyzer):
        print(f"\n=== {analyzer_name} ===")
        stream = analyzer.tokenStream("", StringReader(text))
        term_attr = stream.addAttribute(CharTermAttribute.class_)
        
        stream.reset()
        tokens = []
        while stream.incrementToken():
            tokens.append(term_attr.toString())
        stream.close()
        
        print("Tokens:", tokens)
        print("Token count:", len(tokens))
        return tokens
    
    print("Bengali Text Analysis Comparison")
    print("Original text:", bengali_text)
    
    for name, analyzer in analyzers.items():
        analyze_text(bengali_text, name, analyzer)

compare_bengali_analyzers()
```

```python
import lucene

# Initialize PyLucene VM (uncomment when running)
# lucene.initVM()

from org.apache.lucene.analysis.bn import BengaliAnalyzer
from org.apache.lucene.analysis import CharArraySet
from java.util import Arrays
from java.io import StringReader

def bengali_analysis_with_custom_stopwords():
    # BengaliAnalyzer with default stop words
    default_analyzer = BengaliAnalyzer()
    
    # Create custom stop words
    custom_stop_words = Arrays.asList(["এবং", "হয়", "একটি", "যায়", "যাই"])
    custom_analyzer = BengaliAnalyzer(CharArraySet(custom_stop_words, True))
    
    bengali_text = '''
    বাংলা ভাষা একটি সুন্দর ভাষা 
    আমি প্রতিদিন স্কুলে যাই এবং সেখানে পড়াশোনা করি
    সূর্য পূর্ব দিকে উঠে এবং পশ্চিম দিকে অস্ত যায়
    '''
    
    from org.apache.lucene.analysis.tokenattributes import CharTermAttribute
    
    def analyze_with_analyzer(text, analyzer, analyzer_name):
        print(f"\n=== {analyzer_name} ===")
        stream = analyzer.tokenStream("field", StringReader(text))
        term_attr = stream.addAttribute(CharTermAttribute.class_)
        
        stream.reset()
        tokens = []
        while stream.incrementToken():
            tokens.append(term_attr.toString())
        stream.close()
        
        print("Tokens:", tokens)
        return tokens
    
    print("Bengali Text Analysis with Different Configurations")
    print("Original text:", bengali_text)
    
    analyze_with_analyzer(bengali_text, default_analyzer, "Default BengaliAnalyzer")
    analyze_with_analyzer(bengali_text, custom_analyzer, "Custom StopWords BengaliAnalyzer")

bengali_analysis_with_custom_stopwords()
```

# Hindi Text Analysis

```python
import lucene

# Initialize PyLucene VM (uncomment when running)
# lucene.initVM()

from org.apache.lucene.analysis.hi import HindiAnalyzer
from org.apache.lucene.analysis.standard import StandardAnalyzer
from java.io import StringReader

def hindi_analysis():
    # Using HindiAnalyzer for Hindi-specific analysis
    hindi_analyzer = HindiAnalyzer()
    
    hindi_text = '''
    हिंदी भाषा में एक नमूना पाठ
    मैं स्कूल जाता हूँ, तुम बाजार जाते हो, हम किताब पढ़ते हैं
    सूरज पूर्व में उगता है और पश्चिम में अस्त होता है
    किताब, किताबें, किताबों  
    '''
    
    # Tokenization with HindiAnalyzer
    stream = hindi_analyzer.tokenStream("field", StringReader(hindi_text))
    from org.apache.lucene.analysis.tokenattributes import CharTermAttribute
    term_attr = stream.addAttribute(CharTermAttribute.class_)
    
    stream.reset()
    tokens = []
    while stream.incrementToken():
        tokens.append(term_attr.toString())
    stream.close()
    
    print("Hindi tokens:", tokens)
    print("Total tokens:", len(tokens))

hindi_analysis()
```
