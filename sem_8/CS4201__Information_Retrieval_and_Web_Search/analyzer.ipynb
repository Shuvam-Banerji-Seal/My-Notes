```python
import lucene

lucene.initVM()
```

```python
from java.io import StringReader
```

```python
from org.apache.lucene.analysis.tokenattributes import CharTermAttribute
```

```python
from org.apache.lucene.analysis.standard import StandardAnalyzer
```

```python
test = "The Quick Brown Fox Jumps Over The Lazy Dog"
```

### StandardAnalyzer

```python
analyzer = StandardAnalyzer()
```

```python
stream = analyzer.tokenStream("", StringReader(test))
stream.reset()
```

```python
tokens = []
while stream.incrementToken():
    tokens.append(stream.getAttribute(CharTermAttribute.class_).toString())
print(tokens)
```

### EnglishAnalyzer

```python
from org.apache.lucene.analysis.en import EnglishAnalyzer

analyzer = EnglishAnalyzer()
stream = analyzer.tokenStream("", StringReader(test))
stream.reset()
tokens = []
while stream.incrementToken():
    tokens.append(stream.getAttribute(CharTermAttribute.class_).toString())
print(tokens)
```

### WhitespaceAnalyzer

```python
from org.apache.lucene.analysis.core import WhitespaceAnalyzer

analyzer = WhitespaceAnalyzer()
stream = analyzer.tokenStream("", StringReader(test))
stream.reset()
tokens = []
while stream.incrementToken():
    tokens.append(stream.getAttribute(CharTermAttribute.class_).toString())
print(tokens)
```

### EnglishAnalyzer with given stopword list

```python
from org.apache.lucene.analysis.en import EnglishAnalyzer
from org.apache.lucene.analysis import CharArraySet

test = "The Quick Brown Fox Jumps Over The Lazy Dog. To be or not to be, that's the question."

custom_stops = ['the', 'to', 'or'] 

stop_set = CharArraySet(len(custom_stops), True) # I want to make it case-insensitive.

for w in custom_stops:
    stop_set.add(w)
```

```python
analyzer = EnglishAnalyzer(stop_set)
stream = analyzer.tokenStream("", StringReader(test))
stream.reset()
tokens = []
while stream.incrementToken():
    tokens.append(stream.getAttribute(CharTermAttribute.class_).toString())
print(tokens)
print(test)
```

```python
analyzer = StandardAnalyzer(stop_set)
stream = analyzer.tokenStream("", StringReader(test))
stream.reset()
tokens = []
while stream.incrementToken():
    tokens.append(stream.getAttribute(CharTermAttribute.class_).toString())
print(tokens)
print(test)
```
